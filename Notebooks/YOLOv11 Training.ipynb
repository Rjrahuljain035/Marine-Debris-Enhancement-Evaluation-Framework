```python
import os
import cv2
import numpy as np
import pandas as pd

from skimage.metrics import (
    peak_signal_noise_ratio,
    structural_similarity
)

from scipy.stats import ttest_rel

# ----------------------------------
# UIQM
# ----------------------------------

def calculate_uiqm(img):

    img = img.astype(np.float32)

    r = img[:,:,0]
    g = img[:,:,1]
    b = img[:,:,2]

    rg = r - g
    yb = (r + g)/2 - b

    uicm = np.std(rg) + np.std(yb)

    gray = cv2.cvtColor(
        img.astype(np.uint8),
        cv2.COLOR_RGB2GRAY
    )

    lap = cv2.Laplacian(
        gray,
        cv2.CV_64F
    )

    uism = lap.var()

    uiconm = gray.std()

    uiqm = (
        0.0282*uicm +
        0.2953*uism +
        3.5753*uiconm
    )

    return uiqm


# ----------------------------------
# Locate Raw Image
# ----------------------------------

raw_roots = [

    "YOLO_Dataset/images/train",

    "YOLO_Dataset/images/val"
]

enh_root = "Results/enhanced_images"

results = []


def find_raw(filename):

    for root_dir in raw_roots:

        for root, dirs, files in os.walk(root_dir):

            if filename in files:

                return os.path.join(
                    root,
                    filename
                )

    return None

    


# ----------------------------------
# Process Dataset
# ----------------------------------

enhanced_files = [

    f for f in os.listdir(enh_root)

    if f.lower().endswith(
        (".jpg",".png",".jpeg")
    )
]

print(
    "Total Enhanced Images:",
    len(enhanced_files)
)

for idx, file in enumerate(enhanced_files):

    raw_path = find_raw(file)

    if raw_path is None:

        continue

    enh_path = os.path.join(
        enh_root,
        file
    )

    raw = cv2.imread(raw_path)

    enh = cv2.imread(enh_path)

    if raw is None or enh is None:

        continue

    raw = cv2.cvtColor(
        raw,
        cv2.COLOR_BGR2RGB
    )

    enh = cv2.cvtColor(
        enh,
        cv2.COLOR_BGR2RGB
    )

    enh = cv2.resize(
        enh,
        (raw.shape[1], raw.shape[0])
    )

    psnr = peak_signal_noise_ratio(
        raw,
        enh,
        data_range=255
    )

    ssim = structural_similarity(
        raw,
        enh,
        channel_axis=2,
        data_range=255
    )

    uiqm_raw = calculate_uiqm(raw)

    uiqm_enh = calculate_uiqm(enh)

    results.append([
        file,
        psnr,
        ssim,
        uiqm_raw,
        uiqm_enh
    ])

    if idx % 500 == 0:

        print(
            f"Processed {idx}/{len(enhanced_files)}"
        )

# ----------------------------------
# Save CSV
# ----------------------------------

df = pd.DataFrame(
    results,
    columns=[
        "Image",
        "PSNR",
        "SSIM",
        "UIQM_Raw",
        "UIQM_Enhanced"
    ]
)

df.to_csv(
    "Results/metrics/benchmark_psnr_ssim_uiqm.csv",
    index=False
)

print(df.head())
```

```python
import pandas as pd

df = pd.read_csv(
    "Results/metrics/benchmark_psnr_ssim_uiqm.csv"
)

print("\n===== AVERAGES =====\n")

print(
    "Mean PSNR:",
    round(df["PSNR"].mean(),3)
)

print(
    "Mean SSIM:",
    round(df["SSIM"].mean(),3)
)

print(
    "Mean UIQM Raw:",
    round(df["UIQM_Raw"].mean(),3)
)

print(
    "Mean UIQM Enhanced:",
    round(df["UIQM_Enhanced"].mean(),3)
)

improvement = (
    (
        df["UIQM_Enhanced"].mean()
        -
        df["UIQM_Raw"].mean()
    )
    /
    df["UIQM_Raw"].mean()
)*100

print(
    "UIQM Improvement:",
    round(improvement,2),
    "%"
)
```

```python
import numpy as np

raw = df["UIQM_Raw"]

enh = df["UIQM_Enhanced"]

mean_diff = np.mean(
    enh - raw
)

pooled_sd = np.sqrt(
    (
        raw.std()**2 +
        enh.std()**2
    ) / 2
)

cohen_d = (
    mean_diff /
    pooled_sd
)

print(
    "\nCohen d =",
    round(cohen_d,3)
)
```

```python
from scipy.stats import ttest_rel
from scipy.stats import wilcoxon

t_stat, p_val = ttest_rel(
    df["UIQM_Enhanced"],
    df["UIQM_Raw"]
)

print(
    "Paired t-test p-value:",
    p_val
)

w_stat, p_w = wilcoxon(
    df["UIQM_Enhanced"],
    df["UIQM_Raw"]
)

print(
    "Wilcoxon p-value:",
    p_w
)
```
